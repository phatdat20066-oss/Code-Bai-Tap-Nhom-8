```python
from matplotlib import pyplot as plt
import numpy as np
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 7]
fig, ax = plt.subplots(2, 2, figsize=(5,5))

ax[0,0].plot(x, y,"red")
ax[0,0].set_title("Line Plot")
ax[0,0].set_xlabel("x")
ax[0,0].set_ylabel("y")

ax[0,1].bar(x, y,)
ax[0,1].set_title("Bar Chart")
ax[0,1].set_xlabel("x")
ax[0,1].set_ylabel("Value")

ax[1,0].pie(x)
ax[1,0].set_title("Pie Chart")

ax[1,1].scatter(x,y)
ax[1,1].set_title("Scatter Plot")
ax[1,1].set_xlabel("x")
ax[1,1].set_ylabel("y")

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# Miền cho x^2 và x^3
x = np.linspace(-10, 10, 500)
# Miền cho sqrt(x) chỉ từ 0 đến 10
x_sqrt = np.linspace(0, 10, 300)
y1 = x**2
y2 = x**3
y3 = np.sqrt(x_sqrt)
plt.figure(figsize=(5,5))
plt.plot(x, y1, label="y = x^2")
plt.plot(x, y2, label="y = x^3")
plt.plot(x_sqrt, y3, label="y = √x")
plt.title("Đồ thị các hàm số y = x^2, y = x^3, y = √x")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True) # Hiển thị lưới
plt.legend()
plt.show()
```

```python
thang = np.array([1,2,3,4,5,6])
doanh_thu = np.array([20,25,23,17,28,24])
loi_nhuan = np.array([15,12,13,10,19,17])
plt.figure(figsize=(5,5))
plt.bar(thang,doanh_thu,width=0.6,label="Doanh Thu")
plt.plot(thang,loi_nhuan,marker="o",linewidth=2,label="Loi Nhuan")
plt.xlabel("Tháng")
plt.ylabel("Triệu Đồng")
plt.title("Bảng Doanh Thu Và Lợi Nhuận")
plt.legend()
plt.show()
```
